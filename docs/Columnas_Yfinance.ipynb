```python
import pandas as pd
import yfinance as yf
```

```python
tickers = ["JPM", "BAC", "WFC", "GS", "MS", "C"]

# Descarga de prueba mínima (solo 1 día) con auto_adjust=True
df = yf.download(tickers=tickers, period="1d", auto_adjust=True)

# 1. Ver la lista completa de columnas (MultiIndex)
print("--- COLUMNAS COMPLETAS (MultiIndex) ---")
print(df.columns.tolist())

# 2. Ver las columnas únicas por activo (Nivel métricas)
print("\n--- MÉTRICAS / VARIABLES DISPONIBLES POR BANCO ---")
metricas = df.columns.get_level_values(0).unique().tolist()
print(metricas)

# 3. Ver los tickers reconocidos en las columnas
print("\n--- TICKERS EN LAS CABECERAS ---")
activos = df.columns.get_level_values(1).unique().tolist()
print(activos)
```

```python
# Universo de bancos
tickers = ["JPM", "BAC", "WFC", "GS", "MS", "C"]

# Descarga de muestra (solo últimos 5 días para no saturar)
# group_by='ticker' agrupa primero por Ticker y luego por métrica
df = yf.download(
    tickers=tickers, period="5d", interval="1d", auto_adjust=True, group_by="ticker"
)

print("=" * 70)
print("ÚLTIMO DÍA REGISTRADO POR CADA BANCO")
print("=" * 70)

resumen_ultimo_dia = []

for ticker in tickers:
    # Extraemos los datos del banco y tomamos la última fila disponible
    sub_df = df[ticker].dropna()
    ultimo_registro = sub_df.iloc[-1].copy()

    # Obtenemos la fecha del índice
    fecha_ultimo_dia = sub_df.index[-1].strftime("%Y-%m-%d")

    # Guardamos en una lista para mostrarlo también como tabla resumen
    ultimo_registro["Ticker"] = ticker
    ultimo_registro["Fecha"] = fecha_ultimo_dia
    resumen_ultimo_dia.append(ultimo_registro)

    # Vista detallada banco por banco
    print(f"\nActivo: {ticker} | Fecha de cotización: {fecha_ultimo_dia}")
    print("-" * 50)
    print(f"  • Apertura (Open) : ${sub_df.iloc[-1]['Open']:.2f}")
    print(f"  • Máximo (High)   : ${sub_df.iloc[-1]['High']:.2f}")
    print(f"  • Mínimo (Low)    : ${sub_df.iloc[-1]['Low']:.2f}")
    print(f"  • Cierre (Close)  : ${sub_df.iloc[-1]['Close']:.2f}")
    print(f"  • Volumen (Volume): {int(sub_df.iloc[-1]['Volume']):,}")

# Tabla consolidada final de todos los bancos
print("\n" + "=" * 70)
print("TABLA COMPARATIVA CONSOLIDADA (ÚLTIMO CIERRE)")
print("=" * 70)
df_resumen = pd.DataFrame(resumen_ultimo_dia)
# Reordenamos columnas para lectura limpia
df_resumen = df_resumen[
    ["Ticker", "Fecha", "Open", "High", "Low", "Close", "Volume"]
]
print(df_resumen.to_string(index=False))
```
